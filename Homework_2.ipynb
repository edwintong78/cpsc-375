# Question 3
Consider the billboard dataset that is supplied with the tidyr package which shows the Billboard top 100 song rankings in the year 2000. Apply the Pandas data wrangling methods to answer these questions. For each question, give the code as a single method chain with multiple “verbs” one after the other, beginning with billboard.

Hints:

- To load datasets from R packages, see Slide 13 of Data wrangling - Pandas.
- Write the code to convert the dataset to a tidy table first.


a. Show for each track, how many weeks it spent on the chart


```python
new_billboard = billboard.melt(
id_vars=['artist','track','date.entered'],
value_vars=billboard.columns.drop([ 'artist','track','date.entered']),
var_name='week',
value_name='rank')
new_billboard['week'] = new_billboard['week'].str.replace('wk', '').astype('Int64')

new_billboard.filter(items=['track', 'week']).groupby('track').count()
```

b. List tracks in decreasing order of number of weeks spent on the chart


```python
new_billboard.sort_values(by='week', ascending=False)
```

c. Show for each track, its top rank


```python
new_billboard.groupby('track').agg({'rank': 'min'})
```

d. List tracks in increasing order of its top rank


```python
new_billboard.groupby('track').agg({'rank': 'min'}).sort_values(by='rank',ascending=False).filter(items=['track'])
```

e. Show for each artist, their top rank


```python
new_billboard.groupby('artist').agg({'rank': 'min'})
```

f. List artists in increasing order of their top rank

```python
new_billboard.groupby('artist').agg({'rank': 'min'}).sort_values(by='rank',ascending=False).filter(items=['artist'])
```

g. List tracks that spent more than 35 weeks in the charts


```python
new_billboard.groupby('track').agg({'week': 'max'}).query('week > 35')
```

h. List tracks that spent more than 35 weeks in the charts along with their artists


```python
new_billboard.groupby(['track', 'artist']).agg({'week': 'max'}).query('week > 35')
```

# Question 4
The demographics.csv  file (available in the Tidy data module on Canvas) gives the proportion of a country’s population in different age groups and some other demographic data such as mortality rates and expected lifetime. You can read a CSV file into a DataFrame using Pandas read_csv(), like so: demo = pd.read_csv("demographics.csv")


```python
import pandas as pd

demo = pd.read_csv('/content/demographics.csv')
demo
```

a. The data is not “tidy”. In 2-3 sentences, explain why.


**The data isn't tidy becuase the Series Name column has many variables stored as row values instead of being in their own column. Also, the values, such as YR2015, are stored as column headers. Raw population counts are mixed in with other values, like mortality rates and life expectancy in the same dataset.**

b. Transform the table to tidy data with one country per row.


```python
pd.melt(demo, id_vars=['Country Name', 'Series Name', 'Series Code'], var_name='Year', value_name='Value')
```

c. Add the male/female population numbers together (i.e., ignore sex-related differences). [Hint: You will have to assign for every pair of columns, e.g., .assign(**{'SP.POP.0014.IN': lambda df: df['SP.POP.0014.MA.IN'] + df['SP.POP.0014.FE.IN']})]


```python
tidy_demo = demo.pivot(index=['Country Name', 'Country Code'], columns='Series Code', values='YR2015').reset_index()

for col in tidy_demo.columns:
    if col not in ['Country Name', 'Country Code']:
        tidy_demo[col] = pd.to_numeric(tidy_demo[col], errors='coerce')

tidy_demo = tidy_demo.assign(
    **{
        'SP.POP.0014.IN': lambda df: df['SP.POP.0014.MA.IN'] + df['SP.POP.0014.FE.IN'],
        'SP.POP.1564.IN': lambda df: df['SP.POP.1564.MA.IN'] + df['SP.POP.1564.FE.IN'],
        'SP.POP.65UP.IN': lambda df: df['SP.POP.65UP.MA.IN'] + df['SP.POP.65UP.FE.IN'],
        'SP.POP.80UP.IN': lambda df: df['SP.POP.80UP.MA'] + df['SP.POP.80UP.FE']
    }
)

display(tidy_demo.head())
```

d. Write code to show the top 5 countries with the lowest proportion of the population below 14 years old (i.e., SP.POP.0014.IN/SP.POP.TOTL) [Code, and list of 5 countries]


```python
(
    tidy_demo
    .assign(proportion_below_14=tidy_demo['SP.POP.0014.IN'] / tidy_demo['SP.POP.TOTL'])
    .sort_values(by='proportion_below_14')
)
```
